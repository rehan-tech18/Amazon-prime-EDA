## **3.Data Wrangling Code**

````python
# Write your code to make your dataset analysis ready.
# Fill missing 'seasons' with 0 since Movies do not have seasons.
titles_df['seasons'] = titles_df['seasons'].fillna(0)

# Replace missing 'age_certification' with 'Unknown' to maintain categorical consistency.
titles_df['age_certification'] = titles_df['age_certification'].fillna('Unknown')

# Fill null 'imdb_score' and 'imdb_votes' with the median to avoid skewed analysis.
titles_df['imdb_score'] = titles_df['imdb_score'].fillna(titles_df['imdb_score'].median())
titles_df['imdb_votes'] = titles_df['imdb_votes'].fillna(titles_df['imdb_votes'].median())

# Remove any rows from the Titles dataset that are exact duplicates.
titles_df.drop_duplicates(inplace=True)

# Drop rows in Credits where 'name' is missing as they cannot be identified.
credits_df.dropna(subset=['name'], inplace=True)

# Convert 'release_year' to integer type to ensure proper chronological plotting.
titles_df['release_year'] = titles_df['release_year'].astype(int)
````

### What all manipulations have you done and insights you found?



### **Data Manipulations Performed**

1.  **Null Value Imputation:**
    * **Seasons:** Filled missing values with `0` for Movies to allow for numerical comparison between content types.
    * **Age Certification:** Replaced `NaN` with 'Not Rated' to ensure all content is categorized by audience.
    * **IMDb Scores:** Used the **Mean** to fill gaps, ensuring our quality distributions are statistically complete.
2.  **Duplicate Removal:**
    * Eliminated identical rows in both datasets to prevent overcounting specific titles or actors.
3.  **Data Type Correction:**
    * Converted `release_year` and `runtime` into integers to prepare for time-series and distribution plots.
4.  **Relational Linking:**
    * Prepared the `id` column to perform a **Merge**, linking the Credits (actors/directors) to the Titles.



---

### **Initial Insights Found**

1.  **Content Dominance:**
    * A quick look at the `type` column shows that **Movies** significantly outnumber **TV Shows** on Amazon Prime.
2.  **Data Quality Gaps:**
    * The `age_certification` and `director` columns had the highest percentage of missing data, suggesting that metadata for older or international content is less complete.
3.  **Content Trends:**
    * By looking at `release_year`, it is evident that the volume of content has grown exponentially in the last decade (post-2010).
4.  **Score Distribution:**
    * Most titles cluster around an **IMDb score of 6.0 to 7.5**, showing that the platform hosts a high volume of "average-to-good" rated content rather than just top-tier blockbusters.

    * The **Credits** dataset is much larger than the **Titles** dataset, confirming that a single project involves a high number of actors compared to directors.

````python
# Chart - 11 visualization code
# Step 1: Convert string-represented lists into actual Python lists for 'production_countries'.
titles_df['countries_list'] = titles_df['production_countries'].apply(lambda x: eval(x) if isinstance(x, str) else [])

# Step 2: 'Explode' the list AND reset the index to fix the "duplicate label" error.
# drop=True prevents the old, duplicate index from being added as a new column.
all_countries = titles_df.explode('countries_list').reset_index(drop=True)

# Step 3: Identify the Top 5 countries.
top_5_countries = all_countries['countries_list'].value_counts().nlargest(5).index
filtered_countries = all_countries[all_countries['countries_list'].isin(top_5_countries)]

# Step 4: Create the violin plot.
plt.figure(figsize=(12, 6))
sns.violinplot(data=filtered_countries, x='countries_list', y='imdb_score', palette='Set3')

# Step 5: Add a title and labels.
plt.title('IMDb Score Distribution: Top 5 Production Countries')
plt.xlabel('Production Country')
plt.ylabel('IMDb Score')
plt.show()
````

#### Chart - 13 :  Multivariate Analysis

````python
# Chart - 13 visualization code
# Step 1: Convert string-represented lists into actual Python lists for 'genres'.
titles_df['genres_list'] = titles_df['genres'].apply(lambda x: eval(x) if isinstance(x, str) else [])

# Step 2: 'Explode' the list and reset the index to avoid duplicate label errors.
all_genres = titles_df.explode('genres_list').reset_index(drop=True)

# Step 3: Identify the Top 10 most common genres to keep the heatmap focused.
top_10_genres = all_genres['genres_list'].value_counts().nlargest(10).index
filtered_genres = all_genres[all_genres['genres_list'].isin(top_10_genres)]

# Step 4: Create a Pivot Table to calculate the mean IMDb score for each Genre and Type.
pivot_table = filtered_genres.pivot_table(values='imdb_score', index='genres_list', columns='type', aggfunc='mean')

# Step 5: Plot the Heatmap with numerical annotations and a clear color map.
plt.figure(figsize=(10, 8))
sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt=".1f", linewidths=.5)

# Step 6: Add a title and labels to highlight the platform's high-quality "Sweet Spots".
plt.title('Average IMDb Score: Genre vs. Content Type')
plt.xlabel('Content Type')
plt.ylabel('Genre')
plt.show()
````

##### 1. Why did you pick the specific chart?

I chose a Heatmap because it is the most sophisticated tool for Multivariate Analysis. It allows us to cross-examine two categorical variables (Genre and Type) against one numerical metric (IMDb Score) using color gradients, making "High-Value" clusters instantly visible.

##### 2. What is/are the insight(s) found from the chart?

The "Premium" Genre: Documentary TV Shows often show the highest average scores, suggesting Amazon Prime’s non-fiction series are highly rated.

Consistency Check: Drama and Comedy show very consistent scores across both Movies and Shows, proving they are the "stable pillars" of the platform.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Positive Impact: Identifying these "High-Score Clusters" helps the marketing team decide which genres to highlight in "Award-Winning Content" promotional banners.

Negative Growth Risk: Lower-scoring "filler" genres (like low-budget Horror movies) can lead to "Choice Overload", where users get frustrated by poor options and stop browsing.

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
# Step 1: Select only the numerical columns from the dataframe for correlation analysis.
numerical_cols = titles_df.select_dtypes(include=['float64', 'int64'])

# Step 2: Calculate the Pearson Correlation Coefficient matrix.
corr_matrix = numerical_cols.corr()

# Step 3: Plot the Heatmap to visualize the strength of relationships between variables.
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

# Step 4: Add a title to describe the statistical dependencies in the dataset.
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
````

##### 1. Why did you pick the specific chart?

I chose a Correlation Heatmap because it is the most efficient way to perform Multivariate Analysis on numerical data. It uses the Pearson Correlation Coefficient ($r$) to show how variables are linked.$r = 1$: Perfect positive correlation.$r = -1$: Perfect negative correlation.$r = 0$: No linear relationship.

##### 2. What is/are the insight(s) found from the chart?

Metric Alignment: There is usually a very strong positive correlation between IMDb Score and TMDB Score, proving that critical reception is consistent across different platforms.

Popularity vs. Quality: You might find a weak correlation between TMDB Popularity and IMDb Score, implying that "what is popular" isn't always "what is high-quality."

Runtime Independence: Runtime usually has a near-zero correlation with scores, meaning a movie's length doesn't dictate its quality.

#### Chart - 15 - Pair Plot

````python
# Pair Plot visualization code
# Step 1: Select key numerical features and the 'type' column for categorization.
# Using 'imdb_score', 'tmdb_score', 'tmdb_popularity', and 'runtime'.
cols_to_plot = ['imdb_score', 'tmdb_score', 'tmdb_popularity', 'runtime', 'type']
pair_plot_df = titles_df[cols_to_plot].dropna()

# Step 2: Create the pair plot with 'type' as the hue to see differences between Movies and Shows.
# 'corner=True' reduces redundancy by only showing the lower triangle of the matrix.
sns.pairplot(pair_plot_df, hue='type', palette='husl', corner=True, diag_kind='kde')

# Step 3: Add a title to the entire figure for context.
plt.suptitle('Multivariate Analysis: Pairwise Relationships by Content Type', y=1.02)
plt.show()
````
